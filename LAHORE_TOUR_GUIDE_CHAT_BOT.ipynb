```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

file="/content/lahore_tour_guide_chatbot_data_10k.csv"
df=pd.read_csv(file)

questions = df["User_Question"].fillna("").astype(str)
vectorizer = TfidfVectorizer(
    lowercase=True,
    stop_words="english"
)
question_vectors = vectorizer.fit_transform(questions)

# CHATBOT FUNCTION
def chart_bot(user_question):
    # Convert user's question into TF-IDF vector
    user_vector = vectorizer.transform([user_question])
    # Calculate similarity
    similarity = cosine_similarity(
        user_vector,
        question_vectors
    )[0]
    # Find best matching question
    best_index = similarity.argmax()
    # Get similarity score
    best_score = similarity[best_index]
    # Get answer
    answer = df.iloc[best_index]["Answer"]
    # Get category
    category = df.iloc[best_index]["Category"]
    # Get place
    place = df.iloc[best_index]["Place"]
    # Confidence checking
    if best_score < 0.15:
        return (
            "Sorry, I don't have enough information "
            "about this in my Lahore Tour Guide data."
        )
    # Return result
    return f"""
Category: {category}

Place: {place}

Answer: {answer}

Similarity Score: {best_score:.2f}
"""
# START CHATBOT
print("\n")
print("        WELCOME TO LAHORE TOUR GUIDE CHAT BOT")

print("""
Ask questions about Lahore.

Examples:

- What is Lahore Fort?
- What places should I visit?
- Where can I eat Lahori food?
- What can I see in Lahore in one day?
- Tell me about Badshahi Mosque.
- Where can I shop in Lahore?

Type 'exit' to stop.
""")

while True:
    user_question = input("\nYou: ")
    # Exit chatbot
    if user_question.lower().strip() == "exit":
        print(
            "\nBot: Goodbye! Have a nice day!"
        )
        break
    # Check empty question
    if not user_question.strip():
        print(
            "\nBot: Please enter your question."
        )
        continue

    # Get chatbot answer
    answer = chart_bot(user_question)

    # Display answer
    print(f"\nBot: {answer}")
```
